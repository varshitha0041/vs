```
import gradio as gr
from transformers import pipeline
from PIL import Image

# Load models directly from Hugging Face Cloud Hub
emotion_model = pipeline("image-classification", model="dima806/facial_emotions_image_detection")
sentiment_model = pipeline("sentiment-analysis", model="distilbert-base-uncased-finetuned-sst-2-english")

def analyze_facial_emotion(image):
    if image is None:
        return "Please upload an image or capture a webcam photo."
    pil_img = Image.fromarray(image).convert("RGB")
    results = emotion_model(pil_img)
    return {res['label'].capitalize(): float(res['score']) for res in results}

def analyze_text_sentiment(text):
    if not text.strip():
        return "Please enter a sentence."
    results = sentiment_model(text)[0]
    return {f"Sentiment: {results['label']}": float(results['score'])}

# Build GUI Layout
with gr.Blocks(title="AI Sentiment & Emotion Engine") as app:
    gr.Markdown("# 🧠 Dual AI Sentiment & Facial Emotion Analyzer")

    with gr.Tab("📷 Facial Emotion Classifier"):
        with gr.Row():
            img_in = gr.Image(sources=["webcam", "upload"], type="numpy", label="Face Input")
            img_out = gr.Label(num_top_classes=4, label="Emotion Predictions")
        btn_img = gr.Button("Analyze Emotion", variant="primary")
        btn_img.click(fn=analyze_facial_emotion, inputs=img_in, outputs=img_out)

    with gr.Tab("📝 Text Sentiment Classifier"):
        with gr.Row():
            txt_in = gr.Textbox(lines=3, placeholder="Type sentence or customer feedback...", label="Text Input")
            txt_out = gr.Label(label="Sentiment Output")
        btn_txt = gr.Button("Analyze Text", variant="primary")
        btn_txt.click(fn=analyze_text_sentiment, inputs=txt_in, outputs=txt_out)

if __name__ == "__main__":
    app.launch(share=True) # share=True creates a temporary public URL in Colab
```
