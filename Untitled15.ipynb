```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error,r2_score as r2_error
df=pd.DataFrame({
    'house size':[1000,1200.1500,1800,2000],
    'house price':[50.60,75,90,100]
})
print(df)
x=df[['house size']]
y=df[['house price']]
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
model=LinearRegression()
model.fit(x_train,y_train)
y_predict=model.predict(x_test)
print(y_predict)
prediction=model.predict([[1600]])
print("preicted value for 1600sqft:",prediction)
```
